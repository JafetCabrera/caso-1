# Caso de BIG DATA: Netflix

---

## ¿Qué datos genera?

Netflix genera grandes cantidades de datos cada vez que los usuarios usan la plataforma.

### Tipo de datos:
- Películas y series vistas
- Tiempo de reproducción
- Pausas y avances
- Gustos de los usuarios
- Dispositivos utilizados
- Ubicación geográfica
- Horarios de uso
- Búsquedas realizadas

### Volumen:
Netflix procesa muchísima información diariamente debido a la cantidad de usuarios que tiene.

### Frecuencia:
Los datos se generan en tiempo real mientras las personas usan la plataforma.

---

## ¿Quién los usa?

Los datos son utilizados por:

- Netflix
- Ingenieros y científicos de datos
- Equipos de marketing
- Productores de contenido
- Sistemas de inteligencia artificial

---

## ¿Qué problema resuelve?

El análisis de BIG DATA ayuda a:

- Recomendar películas y series
- Mejorar la experiencia del usuario
- Reducir cancelaciones
- Crear contenido basado en gustos
- Mejorar la calidad del streaming

---

## ¿Por qué no bastaría Excel?

### Volumen
La cantidad de datos es demasiado grande para manejarla en Excel.

### Velocidad
Los datos se generan constantemente y deben analizarse rápido.

### Variedad
Netflix trabaja con diferentes tipos de información como videos, texto y estadísticas.

### Complejidad
Se necesitan tecnologías más avanzadas para analizar tantos datos al mismo tiempo.

---

> Netflix utiliza herramientas de BIG DATA como Hadoop y Spark para procesar toda esta información.
